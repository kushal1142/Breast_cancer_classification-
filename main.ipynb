## Breast Cancer Classificaton with a simple Neural Network 

## Importing the Dependencies

```python
import numpy as np 
import pandas as pd  
import matplotlib.pyplot as plt 
import sklearn.datasets 
from sklearn.model_selection import train_test_split 
```

## Data collection and Preprocessing 

```python
breast_cancer_datasets = sklearn.datasets.load_breast_cancer()
print(breast_cancer_datasets)
```

```python
data_frame = pd.DataFrame(breast_cancer_datasets.data , columns= breast_cancer_datasets.feature_names)
```

```python
data_frame.head()
```

```python
data_frame['label']= breast_cancer_datasets.target
```

```python
x= data_frame.drop(columns= 'label', axis= 1)
y = data_frame['label']
```

```python
print(x)
```

```python
print(y)
```

```python
x_train,x_test , y_train, y_test = train_test_split( x, y , test_size=0.2 , random_state=42)
```

## Standarize the data

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_train_std = scaler.fit_transform(x_train)
x_test_std = scaler.fit_transform(x_test)
```

```python
import tensorflow as tf 
tf.random.set_seed(3)
from tensorflow import keras 
```

```python
model = keras.Sequential([
                          
                          keras.layers.Dense(20, activation='relu'),
                          keras.layers.Dense(2, activation='softmax')
])
```

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
history = model.fit(x_train_std, y_train, validation_split=0.1, epochs=10)
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title('model accuracy')
plt.ylabel('accuracy')
plt.xlabel('epoch')

plt.legend(['training data', 'validation data'], loc = 'lower right')
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title('model loss')
plt.ylabel('loss')
plt.xlabel('epoch')

plt.legend(['training data', 'validation data'], loc = 'upper right')
```

## Accuracy of the model on test data

```python
loss, accuracy = model.evaluate(x_test_std, y_test)
print(accuracy)
```
